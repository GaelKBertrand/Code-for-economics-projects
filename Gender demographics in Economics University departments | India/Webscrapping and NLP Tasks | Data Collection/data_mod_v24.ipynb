```python
import pandas as pd
import re

# Load your dataframe (replace 'your_file.csv' with your actual filename)
#df = pd.read_csv('/content/master_dataset.csv')

def is_not_linkedin(url):
    """Check if the URL is not a valid LinkedIn link."""
    if pd.isna(url):
        return False
    return not re.match(r'^https?://(\w+\.)?linkedin\.com/.*', url, re.IGNORECASE)

# Identify rows where 'linkedin' and 'linkedin_site_manual' contain non-LinkedIn URLs
invalid_linkedin_mask = df['linkedin'].apply(is_not_linkedin)
invalid_linkedin_manual_mask = df['linkedin_site_manual'].apply(is_not_linkedin)
print("Invalid LinkedIn entries found in 'linkedin':", df.loc[invalid_linkedin_mask, 'linkedin'])
print("Invalid LinkedIn entries found in 'linkedin_site_manual':", df.loc[invalid_linkedin_manual_mask, 'linkedin_site_manual'])

# Move invalid 'linkedin' values to 'personal_site_manual' if it's empty
df.loc[invalid_linkedin_mask & df['personal_site_manual'].isna(), 'personal_site_manual'] = df.loc[invalid_linkedin_mask, 'linkedin']
df.loc[invalid_linkedin_manual_mask & df['personal_site_manual'].isna(), 'personal_site_manual'] = df.loc[invalid_linkedin_manual_mask, 'linkedin_site_manual']
print("Updated 'personal_site_manual' with invalid LinkedIn entries.")

# Remove invalid values from 'linkedin' and 'linkedin_site_manual'
df.loc[invalid_linkedin_mask, 'linkedin'] = None
df.loc[invalid_linkedin_manual_mask, 'linkedin_site_manual'] = None
print("Removed invalid LinkedIn entries from 'linkedin' and 'linkedin_site_manual'.")


# Save the cleaned dataframe back to the same file
df.to_csv('/content/master_dataset2025).csv', index=False)
```

```python
import pandas as pd

# Define author name and gender columns
author_name_cols = [f'author_name_{i}' for i in range(1, 8)]
author_gender_cols = [f'author_gender_{i}' for i in range(1, 8)]

# Filter dataset for 'DSE Winter School' duplicates
dse_df = df[df['conf'] == 'DSE Winter School']

# Identify duplicate papers
duplicates = dse_df[dse_df.duplicated(subset=['paper_title'], keep=False)]

# Function to merge duplicate authors and gender columns
def merge_authors(group):
    combined_row = group.iloc[0].copy()  # Start with the first row
    index = 1  # Track position in author columns

    # Loop through each duplicate row
    for _, row in group.iloc[1:].iterrows():
        for i in range(1, 8):  # Loop through author slots 1-7
            if pd.isna(combined_row[f'author_name_{i}']):  # Fill empty slots
                combined_row[f'author_name_{i}'] = row[f'author_name_{i}']
                combined_row[f'author_gender_{i}'] = row[f'author_gender_{i}']

    return combined_row

# Apply merging function to duplicates
merged_dse = duplicates.groupby('paper_title', group_keys=False).apply(merge_authors)

# Drop old duplicates from df and add the merged rows
df = df[~df.index.isin(duplicates.index)]  # Remove old duplicates
df = pd.concat([df, merged_dse], ignore_index=True)  # Add merged rows

# Save updated dataset
df.to_csv('/content/draft-conferences.csv', index=False)

# Display merged data
print("\n✅ Merged duplicate papers for DSE Winter School (Saved to CSV):")
print(merged_dse[['paper_title', 'conf'] + author_name_cols + author_gender_cols].to_string(index=False))
```

```python
import pandas as pd
import re
from gender_guesser.detector import Detector

detector = Detector()

# Function to clean name by removing titles
def clean_name(name):
    if pd.isna(name) or not isinstance(name, str):  # Ensure valid string
        return None
    titles = r"Dr\.|Mr\.|Mrs\.|Ms\.|Miss|Prof\.|Shri\.|Shreemati|Shree"
    return re.sub(titles, "", name).strip()

# Function to categorize gender
def categorize_gender(name):
    if not name:  # If name is empty or None
        return "Not Identified"
    first_name = name.split()[0]  # Extract first name
    gender = detector.get_gender(first_name)
    if gender in ["male", "mostly_male"]:
        return 0
    elif gender in ["female", "mostly_female"]:
        return 1
    else:
        return "Not Identified"

# Sample DataFrame
#data = {"Name": ["Dr. Bijaya Kumar Panda", "Shri. Sangita Roy", "Ashok Parikh", "Prof Abdul Aziz", "Indrani Chakraborty", None]}
#df = pd.DataFrame(data)

# Process names line by line
for i, row in df.iterrows():
    cleaned_name = clean_name(row["PRESIDENT"])  # Clean name
    gender = categorize_gender(cleaned_name)  # Detect gender
    df.at[i, "Gender"] = gender  # Store result in the same row

# Print results line by line
for _, row in df.iterrows():
    print(f"PRESIDENT: {row['PRESIDENT']}, Gender: {row['Gender']}")
```

```python
import pandas as pd
import re
from guess_indian_gender import IndianGenderPredictor

# Initialize Gender Predictor
gp = IndianGenderPredictor()

# Sample DataFrame
#data = {"Name": ["Dr. Bijaya Kumar Panda", "Shri. Sangita Roy", "Ashok Parikh", "Prof Abdul Aziz", "Indrani Chakraborty", None]}
#df = pd.DataFrame(data)

# List of common prefixes to remove
prefixes = r"^(Dr\.|Shri\.|Prof\.|Mr\.|Mrs\.|Miss\.|Ms\.)\s*"

# Function to clean name and remove prefixes
def clean_name(name):
    if pd.isna(name) or not isinstance(name, str) or len(name.strip()) == 0:
        return None
    return re.sub(prefixes, "", name).strip()

# Function to get gender
def get_gender(name):
    if not name:
        return "Not Identified"

    gender = gp.predict(name)

    if gender == "male":
        return 0
    elif gender == "female":
        return 1
    else:
        return "Not Identified"

# Process names
df["Name"] = df["Name"].apply(clean_name)  # Remove prefixes
df["Gender"] = df["Name"].apply(get_gender)  # Get gender

# Print results line by line
for _, row in df.iterrows():
    print(f"Name: {row['Name']}, Gender: {row['Gender']}")
```
