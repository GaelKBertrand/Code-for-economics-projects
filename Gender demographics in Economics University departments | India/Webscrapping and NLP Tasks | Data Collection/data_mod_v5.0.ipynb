### Search operation for institutions' urls

```python
# Function to fetch the college URL using Google Custom Search API
def get_college_url(college_names, service, cse_id, delay=2):
    try:
        if pd.isna(college_names):
            return ""
        print(f"Fetching URL for: {college_names}")
        res = service.cse().list(
            q=college_names,
            cx=cse_id,
        ).execute()
        print(f"Response received for: {college_names}")

        search_items = res.get('items', [])
        if search_items:
            return search_items[0].get('link')
        else:
            return ""
    except Exception as e:
        print(f"Error fetching URL for {college_names}: {e}")
        return ""
    finally:
        time.sleep(delay)

# Build the service
service = build("customsearch", "v1", developerKey=API_KEY)
```

### Search operation for wayback machine urls & snapshots


```python
# Function to save URLs to the Wayback Machine with error handling and exponential backoff
def get_existing_wayback_url(url, retries=5, delay=10):
    """
    Retrieves the URL of the closest existing snapshot from the Wayback Machine for a given URL.

    Args:
    - url (str): The URL to check for existing snapshots.
    - retries (int): Number of times to retry in case of failure (default is 5).
    - delay (int): Delay in seconds between retries (default is 10).

    Returns:
    - str: The URL of the closest existing snapshot, "No snapshot available" if none found,
           or "Error retrieving snapshot" if an error occurs after all retries.
    """
    if not url:
        return ""

    api_url = f"http://archive.org/wayback/available?url={url}"
    for attempt in range(retries):
        try:
            response = requests.get(api_url)
            if response.status_code == 200:
                data = response.json()
                snapshots = data.get("archived_snapshots", {})
                if "closest" in snapshots:
                    return snapshots["closest"]["url"]
                else:
                    return "No snapshot available"
            else:
                print(f"Failed to retrieve snapshot for {url}. Status code: {response.status_code}")
        except requests.RequestException as e:
            print(f"Request exception for {url}: {e}")

        print(f"Retrying in {delay} seconds...")
        time.sleep(delay)

    return "Error retrieving snapshot"
```

### Adding Google Scholar profiles

```python
# Setup logging
logging.basicConfig(filename='scholar_scraper.log', level=logging.INFO)

# Initialize user agent and fuzzy matching
ua = UserAgent()

def get_scholar_profile_url(name, delay=2):
    """
    Retrieves the Google Scholar profile URL for a given name using Google Custom Search API.

    Args:
    - name (str): The name of the scholar to search for.

    Returns:
    - str: The URL of the scholar's profile or "not found" if no profile is found.
    """
    try:
        query = f'"{name}" site:scholar.google.com'
        url = f"https://www.googleapis.com/customsearch/v1?q={query}&cx={CSE_ID}&key={API_KEY}"
        headers = {'User-Agent': ua.random}
        response = requests.get(url, headers=headers)

        if response.status_code == 200:
            results = response.json()
            logging.info(f"Results for {name}: {results}")  # Log the entire response

            if 'items' in results:
                for item in results['items']:
                    if 'scholar.google.com/citations?' in item['link']:
                        return item['link']
                return "not found"
            else:
                return "not found"
        elif response.status_code == 429:
            logging.warning(f"Rate limit hit for {name}. Waiting for 60 seconds before retrying...")
            time.sleep(1)  # Wait for 1 seconds before retrying
            return get_scholar_profile_url(name, delay)
        else:
            logging.error(f"Error fetching Scholar profile URL for {name}: {response.status_code} {response.text}")
            return "not found"
    except Exception as e:
        logging.error(f"Error fetching Scholar profile URL for {name}: {e}")
        return "not found"

```

### Scrapping action




```python
from requests_html import HTMLSession
import pandas as pd
import time

# Load your DataFrame
df = pd.read_csv('/content/drive/My Drive/difference_in_names.csv')  # Update path if needed

# Check if 'scholar_profile_url' exists
if 'scholar_profile_url' not in df.columns:
    raise ValueError("Column 'scholar_profile_url' not found in the DataFrame. Please check the column names.")

# Get the total number of rows
total_rows = df.shape[0]
batch_size = 30

# Initialize the HTML session
session = HTMLSession()

# Prepare lists to store the new data
names = []
total_citation = []
citations_dynmc = []
h_index = []
h_index_5y_dynmc = []
i10_index = []
i10_index_5y_dynmc = []
co_authors_list = []  # List to store co-authors data

# Loop through each batch of URLs
for start in range(0, total_rows, batch_size):
    end = min(start + batch_size, total_rows)
    batch_urls = df['scholar_profile_url'][start:end]

    # Process each URL in the current batch
    for url in batch_urls:
        try:
            # Fetch the URL
            r = session.get(url)

            # Extract scholar profile data
            name = r.html.find('#gsc_prf_in', first=True).text
            citations_all = r.html.find('.gsc_rsb_std', first=True).text
            citations_5y = r.html.find('#gsc_rsb_st > tbody > tr:nth-child(1) > td:nth-child(3)', first=True).text
            h_index_value = r.html.find('#gsc_rsb_st > tbody > tr:nth-child(2) > td:nth-child(2)', first=True).text
            h_index_5y = r.html.find('#gsc_rsb_st > tbody > tr:nth-child(2) > td:nth-child(3)', first=True).text
            i10_index_value = r.html.find('#gsc_rsb_st > tbody > tr:nth-child(3) > td:nth-child(2)', first=True).text
            i10_index_5y = r.html.find('#gsc_rsb_st > tbody > tr:nth-child(3) > td:nth-child(3)', first=True).text

            # Fetch co-authors
            co_authors_elements = r.html.find('#gsc_rsb_co')
            if co_authors_elements:
                co_authors_text = co_authors_elements[0].text
                co_authors = [line.split('\n')[0] for line in co_authors_text.split('\n') if line.strip() and not line.startswith('#')]
                if co_authors and co_authors[0].lower().startswith('co-authors'):
                    co_authors = co_authors[1:]  # Skip the header
                co_authors_list.append(', '.join(co_authors) if co_authors else 'not found')
            else:
                co_authors_list.append('not found')  # No co-authors found

            # Append data to the lists
            names.append(name)
            total_citation.append(citations_all)
            citations_dynmc.append(citations_5y)
            h_index.append(h_index_value)
            h_index_5y_dynmc.append(h_index_5y)
            i10_index.append(i10_index_value)
            i10_index_5y_dynmc.append(i10_index_5y)

            # Print status
            print(f"Processed {name} - {url}")

        except Exception as e:
            print(f"Error fetching data for URL {url}: {e}")
            # Append 'not found' in case of an error for all new columns
            names.append('not found')
            total_citation.append('not found')
            citations_dynmc.append('not found')
            h_index.append('not found')
            h_index_5y_dynmc.append('not found')
            i10_index.append('not found')
            i10_index_5y_dynmc.append('not found')
            co_authors_list.append('not found')

    # Delay between batches
    print(f"Batch {start // batch_size + 1} completed. Taking a 10-second break...")
    time.sleep(10)

# Update the existing columns in the DataFrame without losing other data
df['names'] = names
df['total_citation'] = total_citation
df['citations_dynmc'] = citations_dynmc
df['h_index'] = h_index
df['h_index_5y_dynmc'] = h_index_5y_dynmc
df['i10_index'] = i10_index
df['i10_index_5y_dynmc'] = i10_index_5y_dynmc
df['co-authors'] = co_authors_list  # Adding the co-authors column

# Save the updated DataFrame back to a new CSV file
df.to_csv('/content/drive/My Drive/difference_in_names2.csv', index=False)

print("Finished processing and saved the updated DataFrame.")
```
